```python
import matplotlib.pyplot as plt
import numpy as np

experiments = ['Algorithm 1 -\nSetting 1', 'Algorithm 2 -\nSetting 1', 
               'Algorithm 1 -\nSetting 2', 'Algorithm 2 -\nSetting 2']
mse_a = [10.6284, 10.6284, 10.6191, 10.6191]
mse_a_prime = [10.0066, 10.0046, 9.7772, 9.8369]
cases_added = [0.1424, 0.1443, 0.2697, 0.2274]
p_values = [5.6739e-03, 5.4055e-03, 3.0925e-05, 2.4224e-04]

x = np.arange(len(experiments))
width = 0.35

fig, ax1 = plt.subplots(figsize=(10, 6))

rects1 = ax1.bar(x - width/2, mse_a, width, label="MSE over A", color='lightgray', edgecolor='black')
rects2 = ax1.bar(x + width/2, mse_a_prime, width, label="MSE over A'", color='dimgray', edgecolor='black')

ax1.set_ylabel('10-time 10-fold avg test MSE', fontsize=20)
ax1.set_xticks(x)
ax1.set_xticklabels(experiments, fontsize=18)
ax1.tick_params(axis='y', labelsize=16)
ax1.set_ylim(9.0, 11.5)

ax2 = ax1.twinx()
ax2.plot(x, cases_added, color='black', marker='o', linestyle='dashed', linewidth=2, label='Percentage of Cases Added')
ax2.set_ylabel('Percentage of cases added', fontsize=20)
ax2.tick_params(axis='y', labelsize=16)
ax2.set_ylim(0.0, 1.0)

for i in range(len(experiments)):
    if p_values[i] < 0.05:
        sig_text = 'p < 0.05'
    else:
        sig_text = 'p $\geq$ 0.05'
    
    y_max = max(mse_a[i], mse_a_prime[i])
    ax1.text(x[i], y_max + 0.1, sig_text, ha='center', va='bottom', fontsize=14, fontweight='bold')

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper right', fontsize=17)

plt.tight_layout()
plt.show()
```
